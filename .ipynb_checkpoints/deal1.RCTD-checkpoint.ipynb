```python
spatial_adata = sc.read_h5ad('./bbknn.cnv.h5ad')
```

```python
print(f"Spatial: {spatial_adata.n_obs} beads × {spatial_adata.n_vars} genes")
```

```python
from rctd import SPOT_CLASS_NAMES

class_labels = dict(enumerate(SPOT_CLASS_NAMES))
```

```python
ref_adata = sc.read_h5ad('./scOUT/sc.h5ad')
```

```python
reference = Reference(ref_adata, cell_type_col="celltype")
```

```python
# Use UMI_min=1 for this small tutorial dataset to keep all beads
_config = RCTDConfig(UMI_min=1, UMI_min_sigma=1)
_t0 = time.time()
result_doublet = run_rctd(spatial_adata, reference, mode='doublet', config=_config)
```

```python
t_doublet = time.time() - _t0
print(f'Doublet mode: {t_doublet:.1f}s')
classes, _counts = np.unique(result_doublet.spot_class, return_counts=True)
for cl, ct in zip(classes, _counts):
    print(f'  {class_labels.get(cl, cl)}: {ct}')
```

```python
import numpy as np
full_weights = np.full((spatial_adata.n_obs, len(result_doublet.cell_type_names)), np.nan)
full_weights[result_doublet.pixel_mask] = result_doublet.weights
spatial_adata.obsm["rctd_weights"] = full_weights
```

```python
import pandas as pd
weights_df = pd.DataFrame(
    result_doublet.weights,
    index=spatial_adata.obs_names[result_doublet.pixel_mask],
    columns=result_doublet.cell_type_names,
)
```

```python
weights_df.to_csv('cell.ratio.csv')
```

```python
df  =weights_df
```

```python
import pandas as pd
import numpy as np

# 假设你的数据存储在 DataFrame df 中
# df 的索引是 cell_id，列是细胞类型名称

# 找到每个 cell_id 权重最高的细胞类型（主要类型）
primary_celltype = df.idxmax(axis=1)
primary_weight = df.max(axis=1)

# 找到第二高的细胞类型（次要类型）
# 方法：移除最高值后取最大值
df_without_max = df.apply(lambda x: x.drop(x.idxmax()) if x.notna().sum() > 1 else x, axis=1)
second_celltype = df_without_max.idxmax(axis=1)
second_weight = df_without_max.max(axis=1)

# 创建结果 DataFrame
result_df = pd.DataFrame({
    'cell_id': df.index,
    'primary_celltype': primary_celltype,
    'primary_weight': primary_weight,
    'secondary_celltype': second_celltype,
    'secondary_weight': second_weight
})

```

```python
result_df.to_csv('./STout/rctd_doublet_results.csv', index=False)
```
