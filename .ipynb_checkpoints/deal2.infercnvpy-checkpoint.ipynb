```python
column_names = ["gene_symbol", "chromosome", "start", "end"]  # 示例字段，可根据实际调整

anno = pd.read_csv(
    "/data/T9.data/import/soft/hg38_gencode_v27.txt",
    delim_whitespace=True,
    engine="python",
    header=None,
    names=column_names
)


anno
```

```python
anno_indexed = anno.set_index("gene_symbol")
# 执行 left join：以 adata.var 的 index 为主（即 adata.var_names）
adata.var = adata.var.join(anno_indexed[["chromosome", "start", "end"]], how="left")
adata.var
```

```python
# 检查是否存在重复基因名称，需要去除
duplicates = adata.var_names[adata.var_names.duplicated()]
print(duplicates)
#Index([], dtype='object')
```

### 提取Epi对cnv聚类

```python
import numpy as np
```

```python
#adata_subset = adata[adata.obs['celltype'] == 'Epithelials']
```

```python
adata.obs['celltype'] = adata.obs['celltype'].replace('Tumor', 'Epithelials')
```

```python
adata.obs['celltype'] = adata.obs['celltype'].cat.add_categories('Tumor')
```

```python
# 然后再执行修改
mask = adata.obs['cnv_leiden'].isin(['3','0'])
adata.obs.loc[mask, 'celltype'] = 'Tumor'

# 验证
print(adata.obs['celltype'].value_counts())
```

```python
sc.pl.umap(adata, color="celltype",  show=False)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5), gridspec_kw={"wspace": 0.5})
sc.pl.umap(adata, color="celltype", ax=ax1, show=False)
sc.pl.umap(adata, color="cnv_score", ax=ax2, show=False)
plt.savefig('./STout/cnv.pdf' )
plt.show()
```

```python
mean_cnv = adata.obs.groupby('cnv_leiden')['cnv_score'].mean()
```

```python
group_means = adata.obs.groupby('cnv_leiden')['cnv_score'].mean()

# 筛选均值大于 0.15 的组
high_mean_groups = group_means[group_means > 0.005]

print("=== 均值 > 0.15 的 cnv_leiden 组 ===")
print(high_mean_groups)

# 获取组名列表
high_mean_group_names = high_mean_groups.index.tolist()
```

```python
# 验证
print(adata.obs['cnv_leiden'].value_counts())
```

```python
threshold = 0.029698
count = (adata.obs['cnv_score'] > threshold).sum()
print(f"大于 {threshold} 的数量: {count}")
```

```python
adata.obs['cnv_score'].max()
```

```python
adata.obs["cnv_status"] = "Normal"
adata.obs.loc[adata.obs["cnv_leiden"].isin(["3",'0']), "cnv_status"] = (
    "Tumor"
)
```

```python
# 单独显示tumor细胞的CNV热图
cnv.pl.chromosome_heatmap(adata[adata.obs["cnv_status"] == "Tumor", :] ,figsize= [20,10] ,show=False)
plt.savefig('./STout/tumor.cnv.pdf')
plt.show()
```

```python
allowed_values = ['0','1', '2', '3', '4', '5', '6', '7']

# 将不在允许列表中的值设置为 8
adata.obs['cnv_leiden'] = np.where(
    adata.obs['cnv_leiden'].isin(allowed_values),
    adata.obs['cnv_leiden'],
    '8'
)
```

```python
# 单独显示normal细胞的CNV热图
cnv.pl.chromosome_heatmap(adata[adata.obs["cnv_status"] == "Normal", :],figsize= [20,10] ,show=False)
plt.savefig('./STout/normal.cnv.pdf')
plt.show()
```
